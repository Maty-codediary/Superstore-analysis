```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)
df.dtypes
```

```python
df.boxplot(column=['Sales'])
```

```python
df.head(10)
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
df['Months'] = df['Order Date'].dt.month
sales_by_month = df.groupby('Months')['Sales'].sum().sort_values(ascending=False).head(10)
print(sales_by_month)
```

```python
sales_by_region = df.groupby('Region')['Sales'].sum().sort_values(ascending=False).head(10)
print(sales_by_region)
```

```python
round(df.groupby('Order ID')['Sales'].sum().mean())
```

```python
round(df['Order ID'].duplicated().sum())
```

```python
round(df['Sales'].isna().sum())
```

```python
top_products.sort_values().plot.barh()
plt.title('Top 10 products by revenue')
plt.xlabel('Total sales ($)')
plt.show()
```

```python
sales_by_month.sort_values().plot.bar()
plt.title('Top 10 months by revenue')
plt.ylabel('Total sales ($)')
plt.show()
```

```python
sales_by_region.sort_values().plot.bar()
plt.title('Revenue by region')
plt.ylabel('Total sales ($)')
plt.show()
```
